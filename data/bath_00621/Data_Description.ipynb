# Import the tools to access the DB

```python
import sqlite3
import pprint
pp = pprint.PrettyPrinter(indent=2)
```

# Connect to the DB and print table headings

```python
conn = sqlite3.connect("anon_cad_metadata.db")
c = conn.cursor()
c.execute('SELECT * FROM metadata LIMIT 1')
names = [description[0] for description in c.description]
pp.pprint('metadata table')
pp.pprint(names)
```

# Query Metadata Table

```python
c.execute("SELECT * FROM metadata WHERE team='A' LIMIT 20")
rows = c.fetchall()
pp.pprint(rows)
```
